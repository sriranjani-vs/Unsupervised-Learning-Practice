```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.decomposition import PCA

#Load the Image
image = Image.open("/content/Screenshot 2026-10-21 193601.png").convert("L")

#Convert image into numpy array
img_array = np.array(image)
print("orignal image shape:",img_array.shape)

#Apply PCA
pca = PCA(n_components=0.45)

#Transform Image
compressed_data = pca.fit_transform(img_array)

#Reconstruct Image
reconstructed_image = pca.inverse_transform(compressed_data)
print("Reconstructedimage Shape: ",reconstructed_image.shape)

#Display orignal and compressed images
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img_array,cmap="gray")
plt.title("Original Image")
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(reconstructed_image.reshape(img_array.shape),cmap="gray")
plt.title("Compressed Image")
plt.axis("off")
plt.show()

#Display compression information
print("orignal data size:",img_array.size)
print("compressed data size:",compressed_data.size)
print("Number of principal componets:",pca.n_components_)
print("Explained variance Ratio:",pca.explained_variance_ratio_.sum())
```
